# Data processing and manipulation

```python
# change of dates and month
# variables to use:
# ['emp_length_int'],['term_int'], ['mths_since_earlist_crline'], ['mths_since_issued']
```

```python
# employement lenght
loan_data['emp_length'].unique()
```

```python
loan_data['emp_length'] = loan_data['emp_length'].str.replace('< 1 year', '0 years')
loan_data['emp_length'] = loan_data['emp_length'].str.extract(r'(\d+)').astype('Int64')
```

```python
loan_data['emp_length'].unique()
```

```python
loan_data['emp_length_int'] = loan_data['emp_length'].astype('float64')
```

```python
loan_data['emp_length_int'].unique()
```

```python
# term length of the loan
loan_data['term'].unique()
```

```python
loan_data['term_int'] = loan_data['term'].str.extract(r'(\d+)').astype('Int64')
```

```python
loan_data['term_int'].unique()
```

```python
# earlist time client having the credit line with us
loan_data['earliest_cr_line'].unique()
```

```python
loan_data['earliest_cr_line_date'] = pd.to_datetime(loan_data['earliest_cr_line'],format='%b-%y')
```

```python
loan_data['earliest_cr_line_date']
```

```python
from datetime import date
today = date.today()
print(today)
```

```python
# use the '2017-12-01' as instructed in the course, as the data is old
# np.timedelta64 not supported, changed another way

ref_date = pd.to_datetime('2017-12-01')
loan_data['mths_since_earlist_crline'] = (
    (ref_date.year - loan_data['earliest_cr_line_date'].dt.year)*12 + 
    (ref_date.month - loan_data['earliest_cr_line_date'].dt.month)
    ).astype('float64')
```

```python
loan_data['mths_since_earlist_crline'].describe()
```

```python
# find the negative, why
test_date = loan_data.loc[:,['earliest_cr_line', 'earliest_cr_line_date', 'mths_since_earlist_crline']][loan_data['mths_since_earlist_crline']<0]
```

```python
loan_data.loc[:,['earliest_cr_line', 'earliest_cr_line_date', 'mths_since_earlist_crline']][loan_data['mths_since_earlist_crline']==587]
```

```python
# test if all the negative ones are due to wrongfully get dates in 60s earlier than 69
test_date['earliest_cr_line'].str.extract(r'(\d+)')[0].unique()
```

```python
# confirmed they are all smaller than 69, which is the earliest date, so assign the same date to them 
# to achieve the most recent approach as the data is too old from the date of 2017

loan_data.loc[loan_data['mths_since_earlist_crline']<0, 'mths_since_earlist_crline']=loan_data['mths_since_earlist_crline'].max()
```

```python
loan_data['mths_since_earlist_crline'].describe()
```

```python
pd.qcut(loan_data['mths_since_earlist_crline'],q=4).value_counts().sort_index()
#almost equal in each bucket
```

```python
# issue date manipulation
loan_data['issue_d'].unique()
```

```python
loan_data['issue_d'] = pd.to_datetime(loan_data['issue_d'],format='%b-%y')

# use the same ref date as 2017
ref_date = pd.to_datetime('2017-12-01')
loan_data['mths_since_issued'] = (
    (ref_date.year - loan_data['issue_d'].dt.year)*12 + 
    (ref_date.month - loan_data['issue_d'].dt.month)
    ).astype('float64')
```

```python
loan_data['mths_since_issued'].describe()
```

```python
pd.qcut(loan_data['mths_since_issued'],q=4).value_counts().sort_index()
# similar distribution, with little variation
```

```python
# Data Manipulation on Discrete Variables
# binary variables to change to dummies
```

```python
loan_data['grade'].unique()
```

```python
pd.get_dummies(loan_data['grade'],prefix='grade', dtype=int)
```

```python
loan_data_dummies = [pd.get_dummies(loan_data['grade'],prefix='grade', dtype=int),
                     pd.get_dummies(loan_data['sub_grade'],prefix='sub_grade', dtype=int),
                     pd.get_dummies(loan_data['home_ownership'],prefix='home_ownership', dtype=int),
                     pd.get_dummies(loan_data['verification_status'],prefix='verification_status', dtype=int),
                     pd.get_dummies(loan_data['loan_status'],prefix='loan_status', dtype=int),
                     pd.get_dummies(loan_data['purpose'],prefix='purpose', dtype=int),
                     pd.get_dummies(loan_data['addr_state'],prefix='addr_state', dtype=int),
                     pd.get_dummies(loan_data['initial_list_status'],prefix='initial_list_status', dtype=int)]
```

```python
loan_data_dummies = pd.concat(loan_data_dummies, axis = 1)
```

```python
loan_data = pd.concat([loan_data,loan_data_dummies], axis=1)
```

```python
loan_data.head(10)
```

# Dealing with Missing Values

```python
pd.options.display.max_rows = None
loan_data.isnull().sum()
```

```python
loan_data['total_rev_hi_lim'] = loan_data['total_rev_hi_lim'].fillna(loan_data['funded_amnt'])
```

```python
loan_data['annual_inc']= loan_data['annual_inc'].fillna(loan_data['annual_inc'].mean())
```

```python
loan_data['mths_since_earlist_crline'] = loan_data['mths_since_earlist_crline'].fillna(0)
```

```python
cols = [
    'acc_now_delinq',
    'total_acc',
    'pub_rec',
    'open_acc',
    'inq_last_6mths',
    'delinq_2yrs',
    'emp_length_int'
]

loan_data[cols] = loan_data[cols].fillna(0)
```

# PD Model

```python
# get the target (0 = default, 1 = non-default)
```

```python
loan_data['loan_status'].unique()
```

```python
loan_data['loan_status'].value_counts()/loan_data['loan_status'].count()
```

```python
list_default = ['Charged Off', 'Default','Does not meet the credit policy. Status:Charged Off','Late (31-120 days)']
```

```python
loan_data['good_bad'] = np.where(loan_data['loan_status'].isin(list_default),0,1)
```

# Splitting Data

```python
from sklearn.model_selection import train_test_split
```

```python
input_train, input_test, target_train, target_test = train_test_split(loan_data.drop('good_bad',axis=1),loan_data['good_bad'], test_size=0.2, random_state = 42)
```
